```python
#Shopping Cart System – Using Lists Objective:
```

```python
cart = ["apple", "banana", "mango"]
print(" Initial Cart:", cart)

cart.append("grapes")
cart.remove("banana")
cart.sort()

print(" Final Cart:", cart)
```

```python
#Student Grades – Using Dictionaries Objective:
```

```python
grades = {"Aarav": 85, "Neha": 90, "Kabir": 78}
print(" Initial Grades:", grades)

grades["Meera"] = 92  # Add new student
grades["Kabir"] = 80   # Update grade

print(" Updated Grades:", grades)
```

```python
#Unique Tag Filter – Using Sets Objective:
```

```python
tags = ["python", "coding", "tutorial", "python", "coding"]
unique_tags = set(tags)

print(" Unique Tags:", unique_tags)

unique_tags.add("beginner")
unique_tags.discard("tutorial")

print(" Final Tags:", unique_tags)
```
