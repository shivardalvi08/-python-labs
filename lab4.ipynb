```python
#Salary Calculator – Arithmetic Operators
```

```python
basic_salary = 35000
bonus = 5000
tax = 2500

monthly_salary = basic_salary + bonus - tax
annual_salary = monthly_salary * 12

print(" Monthly Salary:", monthly_salary)
print(" Annual Salary:", annual_salary)
```

```python
# Login Access – Logical & Membership Operators
```

```python
registered_users = ["admin", "manager", "staff"]
username = input("Enter username: ")
password = input("Enter password: ")
access_granted = username in registered_users and password =="admin123"
print("Access Granted:", access_granted)
```

```python
#Product Comparison – Identity & Assignment Operators
```

```python
product_A = 100
product_B = 100
product_C = product_A

product_A += 20
product_B *= 2

print("Product A:", product_A)
print("Product B:", product_B)
print("Product C:", product_C)

print(" A is B:", product_A is product_B)
print(" A is C:", product_A is product_C)
```
