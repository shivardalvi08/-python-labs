```python
#Ticket Pricing System – if, if-else, if-elif-else Objective:
```

```python
age = int(input("Enter your age: "))

if age < 5:
    price = 0
elif age <= 18:
    price = 150
elif age <= 60:
    price = 250
else:
    price = 100

print(f" Your ticket price is ₹{price}")
```

```python
#Login Attempt Tracker – while Loop Objective:
```

```python
correct_password = "secure123"
attempts = 0

while attempts < 3:
    pwd = input("Enter password: ")
    if pwd == correct_password:
        print(" Access Granted")
        break
    else:
        print(" Incorrect password")
        attempts += 1
else:
    print("Account Locked")
```

```python
# Inventory Listing – for Loop with continue/pass Objective:
```

```python
inventory = ["laptop", "defective", "mouse", "pass", "keyboard"]

for item in inventory:
    if item == "defective":
        continue
    elif item == "pass":
        pass  # Placeholder logic
    print(" Item Available:", item)
```

```python
 #Factorial Calculator – for vs while Objective:
```

```python
# Using for loop
num = 5
fact = 1
for i in range(1, num + 1):
    fact *= i
print(" Factorial (for):", fact)
```

```python
# Using while loop
n = 5
fact2 = 1
while n > 0:
    fact2 *= n
    n -= 1
print(" Factorial (while):", fact2)
```
